```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import random
import seaborn as sns
```

```python
marks = []

for i in range(400):
    marks.append(random.randint(35, 95))

df = pd.DataFrame({'Marks': marks})

df['MinMax'] = (df['Marks'] - df['Marks'].min()) / (df['Marks'].max() - df['Marks'].min())

df['Z_std'] = (df['Marks'] - df['Marks'].mean()) / df['Marks'].std()

mad = (df['Marks'] - df['Marks'].mean()).abs().mean()
df['Z_MAD'] = (df['Marks'] - df['Marks'].mean()) / mad

j = len(str(int(df['Marks'].abs().max())))
df['Decimal'] = df['Marks'] / (10 ** j)

print(df)
```

```python
plt.figure(figsize=(10,6))

sns.lineplot(x=df['Marks'], y=df['MinMax'], label='Min-Max Scaling')
sns.lineplot(x=df['Marks'], y=df['Z_std'], label='Z-score (Std)')
sns.lineplot(x=df['Marks'], y=df['Z_MAD'], label='Z-score (MAD)')
sns.lineplot(x=df['Marks'], y=df['Decimal'], label='Decimal Scaling')

plt.xlabel("Data Values (Marks)")
plt.ylabel("Scaled Range")
plt.title("Feature Scaling Techniques Comparison")
plt.grid(True)
plt.show()
```
